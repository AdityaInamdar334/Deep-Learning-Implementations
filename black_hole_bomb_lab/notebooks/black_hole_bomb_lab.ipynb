## 1. Growth vs saturation

The observed early growth rate was 0.081/s and plateau about 3.06 mm. We *choose* a Stuart–Landau normal form to illustrate a finite-amplitude solution. The input seed is arbitrary. A linear prediction does not remain valid at large amplitudes.

```python
t=np.linspace(0,110,1101)
plt.plot(t,linear_amplitude(t)*1e3,label='Linear, invalid at large amplitude')
plt.plot(t,saturated_amplitude(t)*1e3,label='Illustrative nonlinear limit')
plt.axhline(3.06,color='0.5',ls=':',label='Reported 3.06 mm')
plt.axvline(80,color='0.5',ls='--',label='Observed ~80 s')
plt.ylim(0,7);plt.xlabel('Time (s)');plt.ylabel('Amplitude (mm)');plt.legend();plt.show()
```

## 2. Harmonics

If second-harmonic amplitude scales as square of fundamental, its early log growth slope is twice that of the fundamental. The second-harmonic coefficient is an arbitrary teaching choice.

```python
a,h,dc=mode_amplitudes(t)
plt.semilogy(t,a*1e3,label='Fundamental')
plt.semilogy(t,h*1e3,label='Second harmonic A²')
plt.semilogy(t,dc*1e3,label='Static deformation A²')
plt.xlabel('Time (s)');plt.ylabel('Toy amplitudes (mm)');plt.legend();plt.show()
```

## 3. Test an energy-transfer hypothesis

Here we invent two energy reservoirs; the model is **not** the experiment's nonlinear PDE. It is an explicit illustration of the distinction between redistribution and dissipation.

```python
e=coupled_energy(t)
plt.plot(t,e['primary']*1e6,label='Primary proxy energy')
plt.plot(t,e['secondary']*1e6,label='Secondary proxy energy')
plt.xlabel('Time (s)');plt.ylabel('mm² proxy');plt.legend();plt.show()
print('End-state input and secondary loss:',e['injection'][-1],e['dissipation'][-1])
```
