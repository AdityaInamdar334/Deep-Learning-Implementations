# Option-Order Sensitivity Lab
An original, synthetic demonstration inspired by arXiv:2611.09803. **Not** a reproduction of paper measurements.


```python
from option_audit.core import make_toy_questions, toy_audit, identical_options_slot_prior
questions = make_toy_questions(240)
for label, strength in [("position biased", .85), ("reduced bias", .20), ("parallel independent", 0)]:
    result = toy_audit(questions, strength)
    print(label, "flip_rate=", round(result["flip_rate"], 3), "accuracy=", round(result["mean_accuracy"], 3))
```

## Identical-option diagnostic
If every option carries the same content, a content-only scorer should assign equal probability to every slot.


```python
for strength in [.85, 0.0]:
    result = identical_options_slot_prior(12, strength)
    print(strength, result)
```

## Controlled training ablation
A small synthetic model learns from examples whose correct option has an intentionally biased position. Shuffle the *same* rows during training to break the shortcut.


```python
from option_audit.train import make_rows, train, eval_model
rows = make_rows(500)
for shuffle in [False, True]:
    model = train(rows, shuffle_options=shuffle)
    print("shuffle", shuffle, "params", model, "eval", eval_model(model))
```
