# Superfluid Helium-4 Physics Lab: Exploration Notebook

This interactive notebook models the fundamental quantum hydrodynamics of liquid Helium-4:
1. **Thermodynamic Regimes**: Normal boiling at 4.22 K and the Lambda transition at $T_\lambda = 2.1768\text{ K}$.
2. **Condensate vs. Superfluid Density**: Why macroscopic superfluid fraction $\rho_s/\rho \to 100\%$ while single-particle BEC occupancy $n_0 \approx 7.5\%$ (Prisk et al., NIST 2017).
3. **Phonon-Roton Dispersion**: Acoustic phonon branch $E(p) = c_s p$ and local parabolic roton minimum $E(p) = \Delta + \frac{(p - p_0)^2}{2\mu}$.
4. **Landau Critical Velocity**: Minimizing $E(p)/p$ yielding $v_c \approx 58.4\text{ m/s}$ and Doppler-shifted excitation energies.
